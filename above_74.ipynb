```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_style(style="whitegrid")
```

```python
df_gols = pd.read_csv('camp-bras-gols-a-74.csv')
```

```python
df_gols.columns
```

```python
df_gols = pd.DataFrame(df_gols)
```

```python
df_gols.select_dtypes(include=['number']).corr()
```

```python
rodada = df_gols[['rodada']].value_counts(ascending=True)
```

```python
rodada_2 = rodada.sort_index(ascending=True)
```

```python
rodada_2.to_csv('rodada.csv')
```

```python
recorte = df_gols[["minuto"]].value_counts(ascending=True)
```

```python
a74 = recorte.sort_index(ascending=True)
```

```python
a74.to_csv('gols-a-74.csv')
```
