```python

import pandas as pd 
```

```python
import matplotlib.pyplot as plt 
```

```python

df_gols = pd.read_csv("campeonato-brasileiro-gols.csv")
```

```python
df_gols = pd.DataFrame(df_gols)
```

```python
df_gols.columns
```

```python
df_gols["minuto"]
```

```python
df_gols["minuto"].value_counts()
```

```python
df_gols["minuto"].describe()
```

```python
df_gols["minuto"] = pd.to_numeric(df_gols["minuto"], errors='coerce')
```

```python
above_74 = df_gols[df_gols["minuto"] > 74] 
```

```python
above_74
```

```python

above_74.reset_index(drop=True, inplace=True)
```

```python
gols_per_minute = above_74["minuto"].value_counts()
```

```python
gols_per_minute
```

```python
gpm = pd.DataFrame(gols_per_minute)
```

```python
gpm.reset_index(inplace=True)
```

```python
gpm
```

```python
gpm.plot(kind='line')
```
